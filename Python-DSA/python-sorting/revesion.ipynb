# Revising Selection sort


````python
def Selection_sort(nums):
    for i in range(len(nums) - 1):
        min_value = nums[i]
        min_index_j = i
        for j in range(i+1, len(nums)):
            if nums[j] < min_value:
                min_value = nums[j]
                min_index_j = j
        nums[i], nums[min_index_j] = nums[min_index_j],nums[i]

    return nums 




print(Selection_sort([2,1,7,3,4,5]))
````

# Selection Sort — Complexity

### Time

```python
for i in range(n - 1):
    for j in range(i + 1, n):
```

Inner loop runs:

```text
(n-1) + (n-2) + ... + 1
= n(n-1)/2
= O(n²)
```

So:

* **Best:** O(n²)
* **Average:** O(n²)
* **Worst:** O(n²)

Why? It always searches for the minimum in the remaining array.

### Space

Extra variables:

```text
i, j, min_value, min_index_j
```

Fixed number of variables → **O(1)**

The original `nums` array is modified **in-place**, so it isn't extra space.

### Final

```text
Time  → O(n²)
Space → O(1)
```

### Quick Rules

```text
1 loop              → O(n)
2 nested loops      → O(n²)
3 nested loops      → O(n³)

Sequential loops    → ADD
Nested loops        → MULTIPLY

Fixed variables     → O(1) space
New array of size n → O(n) space
```


# insertion Sort


````python
def insertion_sort(nums):

    for i in range(1, len(nums)):

        j = i

        while j > 0:

            if nums[j] < nums[j-1]:

                nums[j], nums[j-1] = nums[j-1], nums[j]

                j -= 1

            else:
                break

    print(nums)


print(insertion_sort([4, 2, 3, 1]))
````
